```python
import torch
import torch.nn as nn
import numpy as np
```

```python
size = np.array([500, 750, 1000, 1250, 1500, 1750, 2000], dtype='float32').reshape(-1, 1)
price = np.array([25, 35, 48, 55, 68, 80, 92], dtype='float32').reshape(-1, 1)
```

```python
size_norm = (size - np.mean(size)) / np.std(size)
size_norm = torch.from_numpy(size_norm)
price = torch.from_numpy(price)
```

```python
model = nn.Linear(in_features=1, out_features=1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
```

```python
for epoch in range(300):
    predicted_price = model(size_norm)
    loss = loss_fn(predicted_price, price)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print('Final loss:', loss.item())
```

```python
new_size = (np.array([[1300]], dtype='float32') - np.mean(size)) / np.std(size)
new_size = torch.from_numpy(new_size)
print('Predicted price for 1300 sq ft:', model(new_size).item(), 'lakhs')
```

```python
import torch
import numpy as np
```

```python
x = torch.arange(7)
print(x)
```

```python
# chunk() - divides a tensor into multiple pieces

a = torch.chunk(x, 4)  # specifying how many parts we want
print(a)
```

```python
# split() - same as chunk() but we can have more control on how we are dividing the tensor

y = torch.arange(5)
b = torch.split(y, [3,2])  # specifying how much large the parts we want
print(b)
```

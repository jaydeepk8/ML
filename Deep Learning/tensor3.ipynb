```python
import torch
import numpy as np
```

```python
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

c = torch.multiply(a,b)
d = torch.matmul(a,b)

print(c)
print(d)
```

```python
x = torch.tensor([
    [1., 2.],
    [3., 4.],
    [5., 6.]
])

print(x)
print(torch.mean(x))
```

```python
print(torch.sum(x))
```

```python
print(torch.std(x))
```

```python
m = torch.tensor([
    [1,2],
    [3,4]
])

n = torch.tensor([
    [5,6],
    [7,8]
])

o = torch.matmul(m,n)
print(o)
```

```python
v = torch.tensor([3., 4.])

print("\nL2 norm:")
print(torch.linalg.norm(v, ord=2))
```
